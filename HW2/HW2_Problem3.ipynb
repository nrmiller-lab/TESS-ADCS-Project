```python
import numpy as np
```

```python
#TESS ADCS Sensors
# two star trackers
# 4 cameras
# 10 coarse sun sensors

#Coarse star tracker specs
sigma_star = 6.0 #arcseconds
sigma_star = sigma_star * 1/3600.0 / 180.0 * np.pi #convert arcseconds to radians
cov_star = sigma_star**2 * np.eye(3)
print("Star tracker covariance matrix: ", cov_star)

#Camera star tracker specs
sigma_cam = 21 * 0.07 #arcsec ######
sigma_cam = sigma_cam * 1/3600.0 / 180.0 * np.pi #convert arcseconds to radians
cov_cam = sigma_cam**2 * np.eye(3)
print("Camera covariance matrix: ", cov_cam)


#Sun sensor specs [Do we want to include? No, It's not really used for pointing]
```

```python
# Simulate Noisy Measurements
#np.random.seed(41)  # for reproducibility
N = 10000  # number of samples

# Assume true measurement is zero vector for simplicity
true_measurement = np.zeros(3)

# Generate noisy measurements
star_measurements = true_measurement + np.random.multivariate_normal(np.zeros(3), cov_star, N)
cam_measurements = true_measurement + np.random.multivariate_normal(np.zeros(3), cov_cam, N)

# --- Verify Statistics ---
print("\nStar Tracker Sample Covariance:\n", np.cov(star_measurements, rowvar=False))
print("Camera Tracker Sample Covariance:\n", np.cov(cam_measurements, rowvar=False))

# Optional: check mean of the simulated measurements
print("\nStar Tracker Sample Mean:", np.mean(star_measurements, axis=0))
print("Camera Tracker Sample Mean:", np.mean(cam_measurements, axis=0))
```
